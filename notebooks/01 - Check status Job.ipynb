Verifica se o Job já roudou com sucesso no mês atual:

```python
import datetime as dt
spark.sql("CREATE TABLE IF NOT EXISTS databricks_cnpj_data_lakehouse.bronze.controle_execucao_job (job_name STRING,  data_ultima_execucao TIMESTAMP)")

mes_atual = dt.datetime.now().strftime("%Y-%m")

df = spark.sql(f"SELECT data_ultima_execucao FROM databricks_cnpj_data_lakehouse.bronze.controle_execucao_job WHERE job_name = 'databricks-cnpj-data-lakehouse-job' AND date_format(data_ultima_execucao, 'yyyy-MM') = '{mes_atual}'")

ja_teve_sucesso = df.count() > 0
deve_rodar = not ja_teve_sucesso

dbutils.jobs.taskValues.set(key="deve_rodar", value=deve_rodar)
print(f"Mês atual: {mes_atual} | Já teve sucesso: {ja_teve_sucesso} | Deve rodar: {deve_rodar}")
```
